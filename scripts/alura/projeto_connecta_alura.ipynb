Qual tipo de interação tem a menor satisfação? E a maior?

```python
df_unificado.groupby('tipo_interacao')['satisfacao_cliente'].mean().round(2).sort_values()
```

Qual canal de atendimento tem a maior duração média? E a maior satisfação?

```python
df_unificado.groupby('canal_interacao').agg(duracao_media=('duracao_minutos', 'mean'), satisfacao_media=('satisfacao_cliente', 'mean')).round(1).sort_values("satisfacao_media", ascending=False)
```

Qual é a taxa de resolução por tipo de interação? As reclamações estão sendo resolvidas?

```python
pd.crosstab(df_unificado['tipo_interacao'], df_unificado['status_resolucao'], normalize='index') * 100
```

Os clientes VIP estão mais satisfeitos com o atendimento do que os clientes Regular?

```python
df_unificado.groupby('segmento_cliente')['satisfacao_cliente'].mean().round(2).sort_values(ascending=False)
```

Qual segmento de cliente gera mais reclamações?

```python
pd.crosstab(df_unificado['segmento_cliente'], df_unificado['tipo_interacao'], normalize='index').round(3) * 100
```

```python
df_analitico['ano_cadastro'] = df_analitico['data_cadastro'].dt.year
```

```python
df_analitico['mes_cadastro'] = df_analitico['data_cadastro'].dt.month
```

```python
df_analitico['dia_semana_interacao'] = df_analitico['data_interacao'].dt.day_name()
```

Existem padrões sazonais nos cadastros de clientes? Em qual período do ano a empresa adquire mais clientes?

```python
df_analitico.groupby('mes_cadastro')['id_cliente'].nunique().sort_index()
```

O WhatsApp está crescendo como canal de atendimento ao longo dos anos?

```python
df_analitico['ano_interacao'] = df_analitico['data_interacao'].dt.year
```

```python
pd.crosstab(df_analitico['ano_interacao'], df_analitico['canal_interacao'], normalize='index').round(2) * 100
```

```python
df_analitico['dias_desde_cadastro'] = (pd.Timestamp("today") - df_analitico['data_cadastro']).dt.days
```

```python
df_analitico.sample(5)
```

```python
df_analitico['perfil_tempo'] = df_analitico['dias_desde_cadastro'].apply(lambda x : 'novo' if x <= 1500 else 'antigo')
```

Distribuição de clientes por segmento

```python
df_analitico['segmento_cliente'].value_counts().plot(kind='pie', title='Distribuição de clientes por segmento', figsize=(4, 4), autopct='%1.1f%%')
```

Duração e satisfação por canal de atendimento

```python
canal_metricas = df_analitico.groupby("canal_interacao").agg(
    duracao_media=("duracao_minutos", "mean"),
    satisfacao_media=("satisfacao_cliente", "mean")
).round(1)
```

```python
canal_metricas['duracao_media'].sort_values().plot(kind='barh', title='Média de duração por canal', figsize=(4, 4), xlabel='minutos', ylabel='', color='red')
```

```python
canal_metricas['satisfacao_media'].sort_values().plot(kind='barh', title='Média de satisfação por canal', figsize=(4, 4), xlim=(0,5), ylabel='')
```

Existe sazonalidade nos cadastros de clientes?

```python
cadastros_por_mes = df_analitico.groupby("mes_cadastro")["id_cliente"].nunique().sort_index()
```

```python
cadastros_por_mes.plot(kind='line', title='Quantidade de clientes por mês', figsize=(4, 4), xlabel='Mês', ylabel='Quantidade de clientes', marker='o', grid=True, color='green')
```
